```python
amount = float(input("Enter amount"))
membership = input("Enter membership_type").Upper()
if amount>500:
    if membership == 'GOLD':
        if amount>500:
           print("Your tota bill is :",(amount - (amount*30/100), "With 30% discount ")
        else :
           print("Your tota bill is :",(amount - (amount*10/100), "With 10% discount ")
    elif  membership == 'SILVER':
        if amount>500:
            print("Your tota bill is :",(amount - (amount*20/100), "With 20% discount ")
        else:
            print("Your tota bill is :",(amount - (amount*5/100), "With 5% discount ")
    else:
            print("Not a valid membership")
else:
     print("No discount on order below 500rs.")
```

```python
amount = float(input("Enter amount: "))
membership = input("Enter membership type: ").upper()

if membership == 'GOLD':
    if amount > 500:
        print("Your total bill is:", amount - (amount * 30 / 100), "With 30% discount")
    else:
        print("Your total bill is:", amount - (amount * 10 / 100), "With 10% discount")
elif membership == 'SILVER':
    if amount > 500:
        print("Your total bill is:", amount - (amount * 20 / 100), "With 20% discount")
    else:
        print("Your total bill is:", amount - (amount * 5 / 100), "With 5% discount")
else:
    print("No discount applied. Total bill:", amount)
```
