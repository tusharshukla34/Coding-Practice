```python
# Identity operators (is, is not) are used to check whether two variables refer to the same object in memory.
```

```python
a = [1,2,3]
b = [1,2,3]
```

```python
print(a is b)
```

```python
b = a
print(a)
print(b)
```

```python
print(a is b)
```

```python
a = [1,2,3,4]
b = [1,2,3,4]
```

```python
print("Id of a is",id(a))
print("Id of b is",id(b))
```

```python
print(a is b)
```

```python
a = [1,2,3,4]
b = a 
```

```python
print("Id of a is",id(a))
print("Id of b is",id(b))
```
