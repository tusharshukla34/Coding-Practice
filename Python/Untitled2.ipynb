```python
Q10. Build a food ordering system. Take order amount and membership type (gold/silver/none) as
input.
Gold member = 30% discount if order above Rs 500, else 10%
Silver member = 20% discount if order above Rs 500, else 5%

No membership = No discount
Print the final amount to pay.
```

```python
for i in range(0,10):
    amount = float(input("Enter the amount of order"))
    mtype = input("Enter silver/Gold/None").lower()
    
    if mtype == "gold":
        if amount > 500:
            print("You got 30% of discount")
            print("After discount you have to pay:",amount - (amount*30/100))
        else:
            print("You got 10% of discount")
            print("After discount you have to pay:",amount - (amount*10/100))
    
    elif mtype == "silver":
        if amount > 500:
            print("You got 20% of discount")
            print("After discount you have to pay:",amount - (amount*20/100))
        else:
            print("You got 5% of discount")
            print("After discount you have to pay:",amount - (amount*5/100))    
    else:
        print("You got NO discount")
        print("You have to pay:",amount)
```

```python
class Student:
    school_name = "Cybrom School"

    def __init__(self,name,roll,marks):
        self.name = name 
        self.roll = roll
        self.marks = marks

    def get_grade(self):

        if self.marks > 90:
            return "A+"
        elif self.marks > 70:
            return "A"
        elif self.marks > 60:
            return "B"
        elif self.marks > 40:
            return "C"
        else:
            return "Fail"

    def show_report(self):

        print("=" * 40)
        print("Your School name is:",self.school_name)
        print("Your Name is:",self.name)
        print("Your Roll no. is:",self.roll)
        print("Your Marks is:",self.marks)
        print("Your Grade is:",self.get_grade())
        print("=" * 40)
```

```python
s1 = Student("tushar",1234,89)    # creating object here
s2 = Student("happy",12345,97)
s3 = Student("jojo",123456,57)
s4 = Student("ramesh",1234567,6)
```

```python
print(s4.show_report())
```
