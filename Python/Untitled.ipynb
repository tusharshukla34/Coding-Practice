```python
while True:
    n = int(input("Enter the number"))
    if n == 1:
        break
    for i in range(0,n):
        print(" * " * (i + 1))
```

```python
num = int(input("Enter the number"))

if num % 2 == 0:
    if num >= 0:
        print("The number is even and +ve")
    else: 
        print("The number is even and -ve")

else:
    if num >= 0:
        print("The number is odd and +ve")
    else: 
        print("The number is odd and -ve")
```
