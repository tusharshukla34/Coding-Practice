```python
Given three sides a, b, c of a triangle, determine ():

First check if it's a valid triangle (sum of any two sides must
be greater than the third). If not → "Invalid Triangle"
If valid, check the type by sides:

All equal → "Equilateral"
Exactly two equal → "Isosceles"
All different → "Scalene"
```

```python
a = 6
b = 5
c = 2

if a+b > c and b+c > a and c+a > b:
    if a == b == c:
        print("Equilateral triangle")
    elif a == b or b == c or c == a:
        print("isoselese")
    else:
        print("Scalene")
else:
    print("Not a triangle")
```
