```python
Write a program that takes an integer input n and outputs:

"Fizz" if n is divisible by 3 but not by 5

"Buzz" if n is divisible by 5 but not by 3

"FizzBuzz" if n is divisible by both 3 and 5

"None" if none of the above conditions are true
```

```python
n = int(input("enter number"))

if n % 3==0:
    if n % 5==0:
        print("FizzBuzz")
    else:
        print("Fizz")
elif n % 5==0:
    print("Buzz")
else:
    print("None")
```
