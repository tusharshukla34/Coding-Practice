```python
# Q7. Take balance and withdrawal amount as input.
# If withdrawal amount is less than or equal to
# balance, print the remaining balance.
# Otherwise print "Insufficient Balance".
```

```python
Q6. Calculate electricity bill based on units consumed.
0 to 100 units = Rs 2 per unit
101 to 300 units = Rs 4 per unit
Above 300 units = Rs 6 per unit
Take units as input and print total bill.
```

```python
unit = float(input("Enter the units"))

if unit >=0 and unit <= 100:
    print("You have used",unit,"units")
    print("--------------------------------")
    print("Your bill is Rs",unit * 2)
elif  unit >=101 and unit <=300:
    print("You have used",unit,"units")
    print("--------------------------------")
    print("Your bill is Rs",unit * 4)
else:
    print("You have used",unit,"units")
    print("--------------------------------")
    print("Your bill is Rs",unit * 6)
```
